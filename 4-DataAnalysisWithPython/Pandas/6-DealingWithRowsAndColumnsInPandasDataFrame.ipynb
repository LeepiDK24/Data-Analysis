# Dealing with Rows and Columns in Pandas DataFrame

```python
df = pd.read_csv("nba.csv")
print(df.head())
```

```python
print(df[['Name' , 'Age']])
```

### Adding Columns
New columns can be added by assigning a list or series to a column name.

```python
df['Experience'] = 1
```

```python
df
```

### Deleting Columns
You can delete a column from the DataFrame using the drop() method.

```python
df.drop('Experience' , axis=1 , inplace=True)
```

```python
df
```

# Dealing with Rows

### Selecting Rows
Rows can be selected using .loc[] by row label or index.

```python
data = pd.read_csv("nba.csv", index_col="Name")

first = data.loc["Avery Bradley"]
second = data.loc["R.J. Hunter"]

print(first, "\n\n", second)
```

### Adding Rows
New rows can be added using pd.concat().

```python
new_row = {
    'Name': 'Geeks',
    'Team': 'Boston',
    'Number': 3,
    'Position': 'PG',
    'Age': 33,
    'Height': '6-2',
    'Weight': 189,
    'College': 'MIT',
    'Salary': 99999
}
```

```python
df = pd.concat([df,pd.DataFrame([new_row])] , ignore_index=True)
df.tail()
```

### Deleting Rows
Rows can be removed using drop() with row labels.

```python
data = pd.read_csv("nba.csv", index_col="Name")
data.drop(["Avery Bradley", "John Holland", "R.J. Hunter"], inplace=True)
```
