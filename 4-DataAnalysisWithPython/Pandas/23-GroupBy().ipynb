```python
import numpy as np
import pandas as pd
```

# Pandas GroupBy
The groupby() function in Pandas is important for data analysis as it allows us to group data by one or more categories and then apply different functions to those groups. This technique is used for handling large datasets efficiently and performing operations like aggregation, transformation and filtration on grouped data.

## Step 1: Splitting Data into Groups

```python
# 1. Group data by a single key
data1 = {'Name':['Jai', 'Anuj', 'Jai', 'Princi',
                 'Gaurav', 'Anuj', 'Princi', 'Abhi'],
        'Age':[27, 24, 22, 32,
               33, 36, 27, 32],
        'Address':['Nagpur', 'Kanpur', 'Allahabad', 'Kannuaj',
                   'Jaunpur', 'Kanpur', 'Allahabad', 'Aligarh'],
        'Qualification':['Msc', 'MA', 'MCA', 'Phd',
                         'B.Tech', 'B.com', 'Msc', 'MA']}
df = pd.DataFrame(data1)
print(df)
print(df.groupby('Name').groups)
```

```python
gk = df.groupby('Name') 
gk.first()
```

```python
# 2. Grouping data with multiple keys
a = df.groupby(['Name','Qualification'])
print(a.groups)
```

```python
# 3. Grouping data by sorting keys: Group keys are sorted by default using the groupby operation.
df.groupby('Name')['Age'].sum()
```

```python
# 4. Grouping data with object attributes Grouping data with object attributes in Pandas allows us to treat the groups 
# attribute as a dictionary where the keys are unique group values
#  and values are the indices (row labels) corresponding to each group.
df.groupby('Name').groups
```

```python
# 5. Iterating through groups
grp = df.groupby('Name')
for name, group in grp:
    print(name)
    print(group)
    print()
```

```python
# 6. Selecting a group
grp = df.groupby('Name')
grp.get_group('Jai')
```

```python
grp = df.groupby(['Name', 'Qualification'])
grp.get_group(('Jai', 'Msc'))
```

## Step 2: Applying Functions to Groups

#### 1. Aggregation:

```python
grp1 = df.groupby('Name')
grp1['Age'].aggregate(np.sum)
```

```python
grp1 = df.groupby(['Name', 'Qualification'])
grp1['Age'].aggregate(np.sum)
```

```python
grp = df.groupby('Name')
grp['Age'].agg([np.sum, np.mean, np.std])
```

#### 2. Transformation:

```python
data2 = {'Name':['Jai', 'Anuj', 'Jai', 'Princi', 
                 'Gaurav', 'Anuj', 'Princi', 'Abhi'], 
        'Age':[27, 24, 22, 32, 
               33, 36, 27, 32], 
        'Address':['Nagpur', 'Kanpur', 'Allahabad', 'Kannuaj',
                   'Jaunpur', 'Kanpur', 'Allahabad', 'Aligarh'], 
        'Qualification':['Msc', 'MA', 'MCA', 'Phd',
                         'B.Tech', 'B.com', 'Msc', 'MA'],
        'Score':[23, 34, 35, 45, 47, 50, 52, 53]} 
        
df2 = pd.DataFrame(data2)
grp2 = df2.groupby('Name')
sc = lambda x: (x - x.mean()) / x.std() 
grp2['Score'].transform(sc)
```

#### 3. Filtration:

```python
grp2 = df2.groupby('Name')
grp2.filter(lambda x: len(x) >= 2)
```

## Step 3: Combining Results

```python
df.groupby('Name').agg({'Age': 'sum'})
```
