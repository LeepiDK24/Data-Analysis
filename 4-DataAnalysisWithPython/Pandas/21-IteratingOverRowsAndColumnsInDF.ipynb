# Iterating over rows and columns in Pandas DataFrame

## Iterate Over Rows

### 1. Using iterrows()

````python
df = pd.DataFrame({
    "Drama": ["GAP", "23.5", "The Loyal Pin"],
    "Couple": ["FreenBecky", "MilkLove", "EngLot"],
    "Episodes": [12, 13, 16],
    "Rating": [9.2, 8.8, 9.0]
})

print(df)
````

````python
for index,row in df.iterrows():
    if row['Rating'] >= 9:
        print(index,row)
````

### 2. Using itertuples()

#### What is `itertuples()`?

`itertuples()` is used to **iterate through a Pandas DataFrame row by row**.

Each row is returned as a **named tuple**.

Basic Syntax

```python
for row in df.itertuples():
    print(row)

Example Output

```text
Pandas(Index=0, Drama='GAP', Couple='FreenBecky', Episodes=12, Rating=9.2)

┌─────────────────────────────────────────────┐
│ Index    → 0                                │
│ Drama    → GAP                              │
│ Couple   → FreenBecky                       │
│ Episodes → 12                               │
│ Rating   → 9.2                              │
└─────────────────────────────────────────────┘

````python
df = pd.DataFrame({
    "Drama": ["GAP", "23.5", "The Loyal Pin"],
    "Couple": ["FreenBecky", "MilkLove", "EngLot"],
    "Episodes": [12, 13, 16],
    "Rating": [9.2, 8.8, 9.0]
})

print(df)
````

````python
result = []

for row in df.itertuples():

    if row.Rating >= 9:
        result.append(row.Drama)

print(result)
````

````python
# Question:
#Find the highest-rated drama having at least 12 episodes.

highest_rated = 0
best_Drama = ""

for row in df.itertuples():
    if row.Episodes >= 12 and row.Rating > highest_rated:
        highest_rated = row.Rating
        best_Drama = row.Drama

print(best_Drama,highest_rated)
````

## Iteraing over Columns

### 1.Pandas `items()`

`items()` is used to **iterate through a DataFrame column by column**.

Each iteration gives:

- `label` → column name
- `values` → column as a Pandas Series

### Syntax

```python
for label, values in df.items():
    print(label)
    print(values)

````python
for label, values in df.items():
    print(label)
    print(values)
````

````python
for label, values in df.items():

    if values.dtype != "object":
        print(label, "→", values.mean())
````

### 2. Using Direct Column Iteration

We can loop directly through the **column names** using `df.columns`.

Syntax

```python
for col in df.columns:
    print(col)

````python
for col in df.columns:
    print(col, "→", df[col])
````
