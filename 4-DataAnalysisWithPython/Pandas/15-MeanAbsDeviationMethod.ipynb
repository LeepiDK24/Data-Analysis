# Mean Absolute Deviation (MAD)

The formula for Mean Absolute Deviation is:

$$
\boxed{MAD = \frac{1}{n}\sum_{i=1}^{n}|x_i-\bar{x}|}
$$

Where:

- $x_i$ = Each individual value
- $\bar{x}$ = Mean of all values
- $n$ = Number of values
- $|x_i-\bar{x}|$ = Absolute deviation of each value from the mean

**In Pandas:**

```python
(df - df.mean()).abs().mean()

### DataFrame Examples

````python
df = pd.DataFrame({"A":[12, 4, 5, None, 1],
                   "B":[7, 2, 54, 3, None],
                   "C":[20, 16, 11, 3, 8], 
                   "D":[14, 3, None, 2, 6]})
````

````python
(df - df.mean(skipna=True)).abs().mean(skipna=True)
````

### Series Examples

````python
s = pd.Series([12, 4, 5, None, 1])
````

````python
(s - s.mean(skipna=True)).abs().mean(skipna=True)
````
