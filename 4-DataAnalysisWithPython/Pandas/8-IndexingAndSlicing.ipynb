# Indexing and Selecting Data with Pandas
There are basically 3 ways of Indexing and Selecting data in Pandas
- Indexing Data using the [] Operator
- Indexing with .loc[ ]
- Indexing with .iloc[]


## Indexing Data using the [] Operator

### 1. Selecting a Single Column

```python
df = pd.read_csv('nba.csv')
col1 = df['Age']
col1
```

### 2. Selecting Multiple Columns

```python
cols = df[['Age','Name','Salary']]
cols
```

## Indexing with .loc[ ]

### 1. Selecting a Single Row by Label

```python
df1 = pd.read_csv('nba.csv' ,index_col='Name')
print(df1.loc['Avery Bradley'])
```

### 2. Selecting Multiple Rows by Label

```python
rows = df1.loc[["Avery Bradley", "R.J. Hunter"]]
print(rows)
```

### 3. Selecting Specific Rows and Columns

```python
df1
```

```python
sel = df1.loc[["Avery Bradley", "R.J. Hunter"],["Team", "Number", "Position"]]
print(sel)
```

### 4. Selecting All Rows and Specific Columns

```python
all_rows_specific_columns = df1.loc[:, ["Team", "Position", "Salary"]]
print(all_rows_specific_columns)
```

## Indexing with .iloc[]

### 1. Selecting a Single Row by Position

```python
data = pd.read_csv("nba.csv", index_col="Name")
row = data.iloc[3]
print(row)
```

### 2. Selecting Multiple Rows by Position

```python
rows = data.iloc[[3, 5, 7]]
print(rows)
```

### 3. Selecting Specific Rows and Columns by Position

```python
selection = data.iloc[[3, 4], [1, 2]]
print(selection)
```

### 4. Selecting All Rows and Specific Columns by Position

```python
selection = data.iloc[:, [1, 2]]
print(selection)
```
