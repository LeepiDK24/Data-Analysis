# Pandas DataFrame drop() Method Syntax
## `DataFrame.drop()`

### Syntax

```python
DataFrame.drop(
    labels=None,
    axis=0,
    index=None,
    columns=None,
    level=None,
    inplace=False,
    errors='raise'
)


| Parameter | Description |
| --- | --- |
| **`labels`** | String or list of strings referring to row or column names. |
| **`axis`** | `0` or `'index'` → Rows; `1` or `'columns'` → Columns. |
| **`index`** | Single label or list of labels referring to rows. Alternative to `axis=0`. |
| **`columns`** | Single label or list of labels referring to columns. Alternative to `axis=1`. |
| **`level`** | Specifies the level when the DataFrame has a multi-level index. |
| **`inplace`** | If `True`, modifies the original DataFrame. |
| **`errors`** | `'raise'` → raises an error if a label doesn't exist; `'ignore'` → ignores missing labels and drops the remaining values. |

### Dropping Rows in Pandas by Index Label

````python
df = pd.read_csv('nba.csv' , index_col='Name')
df
````

````python
df.drop(['John Holland' , 'R.J. Hunter' , 'Raul Neto'] , inplace=True)
df
````

### Dropping Columns in Pandas with Column Name

````python
df.drop(['Position' , 'Number'] , axis=1 , inplace=True)
df
````
