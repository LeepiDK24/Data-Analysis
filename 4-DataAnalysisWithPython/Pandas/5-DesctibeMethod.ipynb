# Pandas DataFrame describe() Method
The describe() method in Pandas generates descriptive statistics of DataFrame columns which provides key metrics like mean, standard deviation, percentiles and more. It works with numeric data by default but can also handle categorical data which offers insights like the most frequent value and the number of unique entries

### `.describe()` - Summary of Data

`.describe()` gives us a **quick statistical summary** of a DataFrame.

### Syntax

> `DataFrame.describe(percentiles=None, include=None, exclude=None)`

### Parameters

- **`percentiles`** → Tells Pandas which percentiles you want.
  - Default: `25%`, `50%`, and `75%`
  - Example: `[0.1, 0.5, 0.9]`

- **`include`** → Tells Pandas which **data types** to include.
  - Example: `include="object"` → only text columns
  - Example: `include="all"` → include all columns

- **`exclude`** → Tells Pandas which **data types** to leave out.
  - Example: `exclude="object"` → exclude text columns

### Easy Way to Remember

> **`percentiles` → Which percentages? 📊**
>
> **`include` → What should I include? ✅**
>
> **`exclude` → What should I leave out? ❌**



### 1. Using describe() on a DataFrame

```python
data = pd.read_csv('nba.csv')
df = pd.DataFrame(data)
print(df.describe())
```

### 2. Customizing describe() with Percentiles

```python
p = [0.20,0.40,0.60,0.80]
i = ['object' , 'float' , 'int']

res = df.describe(percentiles=p , include=i)
print(res)
```

### 3. Describing String (Object) Data
For string (object) data, describe() provides:

- count: Total number of non-null values.
- unique: Number of unique values in the column.
- top: Most frequent value in the column.
- freq: Frequency of the most common value.

```python
res = df['Name']
print(res.describe())
```

### 4. Describing Specific Columns with describe()

```python
salary = df['Salary']
print(salary.describe())
```

### 5. Describing Data with include='all'
By using the include='all' parameter we can generate a summary for all columns in the DataFrame regardless of data type. 

```python
res = df.describe(include='all')
print(res)
```
