# Adding New Column to Existing DataFrame in Pandas
There are 5 different ways to do it which are as follows :

- By Declaring a New List as column
- By using DataFrame.assign() method
- By using Dictionary
- By using DataFrame.insert()
- By using Dataframe.loc()

### By Declaring a New List as column

```python
data = {'Name': ['Pandas', 'Geeks', 'for', 'Geeks'],
        'Height': [1, 2, 3, 4],
        'Qualification': ['A', 'B', 'C', 'D']}

df = pd.DataFrame(data)
address = ['NewYork', 'Chicago', 'Boston', 'Miami']
df['Address'] = address
df
```

### By using DataFrame.assign() method

```python
df = df.assign(address = ['Mumbai', 'Delhi', 'Chenai', 'Hyd'])
df
```

### By using Dictionary

```python
data = {'Name': ['Pandas', 'Geeks', 'for', 'Geeks'],
        'Height': [1, 2, 3, 4],
        'Qualification': ['A', 'B', 'C', 'D']}

df = pd.DataFrame(data)

address = {'Pandas': 'NewYork', 'Geeks': 'Chicago', 
            'for': 'Boston', 'Geeks_2': 'Miami'}
df
```

```python
# Method -1
df['Address'] = df['Name'].map(address)
df
```

```python
# Method-2
new_columns = {'Age': [21, 22, 23, 24], 
               'City': ['NY', 'LA', 'SF', 'DC']}

df = df.assign(**new_columns)
df
```

### By using DataFrame.insert()

```python
data = {'Name': ['Pandas', 'Geeks', 'for', 'Geeks'],
        'Height': [1, 2, 3, 4],
        'Qualification': ['A', 'B', 'C', 'D']}

df = pd.DataFrame(data)

df.insert(2,"Age",[21,22,22,12],True)
df
```

### By using Dataframe.loc()

```python
data = {'Name': ['Pandas', 'Geeks', 'for', 'Geeks'],
        'Height': [1, 2, 3, 4],
        'Qualification': ['A', 'B', 'C', 'D']}

df = pd.DataFrame(data)
df.loc[df['Height'] >=3, 'Category'] = 'Tall'
df.loc[df['Height'] <3, 'Category'] = 'Short'
df
```
