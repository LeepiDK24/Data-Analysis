# Examples of Extracting rows using .loc[]

### 1. Multiple parameters

```python
rows = df.loc[['Jae Crowder' , 'Raul Neto']]
rows
```

### 2. Extracting multiple rows with same index

```python
df = pd.read_csv('nba.csv' ,index_col='Team')
```

```python
rows = df.loc["Utah Jazz"]
rows
```

### 3. Extracting rows between two index labels

```python
data = pd.read_csv("nba.csv", index_col ="Name") 
```

```python
rows = data.loc['Avery Bradley' : 'Jonas Jerebko']
rows
```

# Extracting rows using Pandas .iloc[] in Python

### 1. Selecting rows using Pandas .iloc

```python
data = pd.read_csv('nba.csv')
row2 = data.iloc[3]
print(row2)
```

### 2.Selecting Multiple Rows using Pandas .iloc[] in Python


```python
data.iloc[[4,5,6,7,8,9]]
```

```python
data.iloc[4:10]
```

### 3.Select Rows by Name or Index usingPandas .iloc[] in Python

```python
data2 = pd.DataFrame({
    'Name': ['Geek1', 'Geek2', 'Geek3', 'Geek4', 'Geek5'],
    'Age': [25, 30, 22, 35, 28],
    'Salary': [50000, 60000, 45000, 70000, 55000]
})
```

```python
data2.set_index('Name' ,inplace=True)
```

```python
row_1 = data2.iloc[0,:]
row_1
```

```python
rows2_to_rows3 = data2.iloc[1:3 ,:]
rows2_to_rows3
```
