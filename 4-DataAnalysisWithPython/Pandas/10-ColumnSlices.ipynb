```python
import pandas as pd
```

```python
df1 = pd.DataFrame({"a": [1, 2, 3, 4, 5, 6, 7], 
                    "b": [2, 3, 4, 2, 3, 4, 5], 
                    "c": [3, 4, 5, 2, 3, 4, 5], 
                    "d": [4, 5, 6, 2, 3, 4, 5], 
                    "e": [5, 6, 7, 2, 3, 4, 5]})
```

```python
df1
```

## Method 1: Slice Columns in pandas using reindex

```python
df2 = df1.reindex(columns=['c','e'])
print(df2)
```

## Method 2: Slice Columns in pandas using loc[]
Syntax: [ : , first : last : step]

```python
df3 = df1.loc[:,'b':'e' : 2]
df3
```

```python
df3 = df1.loc[:,'b':'e' : 1]
df3
```

## Method 3: Slice Columns in pandas using iloc[]
Syntax: [ : ,start : stop : step]

```python
df2 = df1.iloc[:, 1:3:1]
print(df2)
```
