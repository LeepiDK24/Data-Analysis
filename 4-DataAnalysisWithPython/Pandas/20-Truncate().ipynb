# Pandas DataFrame.truncate
### Syntax

```python
DataFrame.truncate(
    before=None,
    after=None,
    axis=None,
    copy=True
)

### Parameters

| Parameter | Description |
| --- | --- |
| **`before`** | Truncate all rows/columns before this index value. |
| **`after`** | Truncate all rows/columns after this index value. |
| **`axis`** | Specifies the axis to truncate. `0` or `'index'` → Rows; `1` or `'columns'` → Columns. |
| **`copy`** | If `True`, returns a copy of the truncated DataFrame. |

````python
df = pd.DataFrame({
    'Drama': ['Love in the Air', 'GAP', 'Blank', '23.5', 'The Loyal Pin'],
    'Actor': ['Boss', 'Freen', 'Faye', 'Milk', 'Becky'],
    'Rating': [9.2, 9.0, 8.8, 8.7, 9.1]
})

index_ = pd.date_range(
    '2024-01-01 18:00',
    periods=5,
    freq='h'
)

df.index = index_
df
````

````python
res = df.truncate(before='2024-01-01 19:00:00' , after='2024-01-01 21:00:00')
print(res)
````

````python
df = pd.DataFrame({
    "Drama": ["GAP", "Blank", "23.5", "The Loyal Pin", "Love in the Air"],
    "Actor": ["Freen", "Faye", "Milk", "Becky", "Boss"],
    "Rating": [9.0, 8.8, 8.7, 9.1, 9.2]
})

index_ = ['Drama_1', 'Drama_2', 'Drama_3', 'Drama_4', 'Drama_5']

df.index = index_

print(df)

result = df.truncate(
    before='Drama_3',
    after='Drama_4'
)
print("\n After Truncating \n")
print(result)
````
