## 1. Accessing a DataFrame with a Boolean Index

A Boolean index means the **index labels of the DataFrame are `True` or `False`**.

````python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Freen", "Becky", "Lingling", "Orm"],
    "Series": ["GAP", "GAP", "The Secret of Us", "The Secret of Us"],
    "Rating": [9.2, 9.5, 9.7, 9.0]
}, index=[True, False, True, False])

print(df)
````

````python
df.loc[True]
````

## 2. Applying a Boolean Mask to a DataFrame

A **Boolean mask** is a collection of `True` and `False` values used to filter rows.

- `True` → keep the row
- `False` → remove the row

````python
df = pd.DataFrame({
    "Name": ["Freen", "Becky", "Lingling", "Orm"],
    "Series": ["GAP", "GAP", "The Secret of Us", "The Secret of Us"],
    "Rating": [9.2, 9.5, 9.7, 9.0]
})

print(df)
````

````python
mask = df["Rating"] > 9.2
````

````python
df[mask]
````

````python
df[df["Rating"] > 9.2]
````

# 3. Masking Data Based on Column Value

Column-based masking means **filtering rows based on the values present in a column**.

Pandas checks each value in the column against a condition and produces `True` or `False`.

- `True` → keep the row
- `False` → remove the row

---

## Basic Syntax

```python
df[df["column"] condition]

````python
# Rating greater than 9
df[df["Rating"] > 9]
````

````python
# Rating 9.5 or higher
df[df["Rating"] >= 9.5]
````

````python
# Rating less than 9
df[df["Rating"] < 9]
````

````python
# Rating 9 or lower
df[df["Rating"] <= 9]
````

````python
# Rating exactly 9.5
df[df["Rating"] == 9.5]
````

````python
# Series is GAP
df[df["Series"] == "GAP"]
````

````python
# Series is NOT GAP
df[df["Series"] != "GAP"]
````

````python
# GAP AND rating above 9.3
df[(df["Series"] == "GAP") & (df["Rating"] > 9.3)]
````

````python
# GAP OR 23.5
df[(df["Series"] == "GAP") | (df["Series"] == "23.5")]
````

````python
# NOT GAP
df[~(df["Series"] == "GAP")]
````

### Most Important Things to Remember

| Symbol / Syntax | Meaning | Example |
|---|---|---|
| `df[condition]` | Filters rows | `df[df["Rating"] > 9]` |
| `==` | Equal to | `df[df["Series"] == "GAP"]` |
| `!=` | Not equal to | `df[df["Series"] != "GAP"]` |
| `>` | Greater than | `df[df["Rating"] > 9]` |
| `<` | Less than | `df[df["Rating"] < 9]` |
| `>=` | Greater than or equal to | `df[df["Rating"] >= 9]` |
| `<=` | Less than or equal to | `df[df["Rating"] <= 9]` |
| `&` | AND — both conditions must be True | `(condition1) & (condition2)` |
| `\|` | OR — at least one condition must be True | `(condition1) \| (condition2)` |
| `~` | NOT — reverses True/False | `~(condition)` |

### Important Rules

- Always put each condition inside `()` when combining conditions.

```python
df[(df["Rating"] > 9) & (df["Series"] == "GAP")]

# 4. Masking Data Based on an Index Value

Index-based masking means **filtering rows based on the values of the DataFrame's index**.

### Basic Syntax

```python
df[df.index condition]

### Comparison Operators

| Operator | Meaning | Example |
|---|---|---|
| `==` | Equal to | `df[df.index == 2]` |
| `!=` | Not equal to | `df[df.index != 2]` |
| `>` | Greater than | `df[df.index > 1]` |
| `<` | Less than | `df[df.index < 2]` |
| `>=` | Greater than or equal to | `df[df.index >= 2]` |
| `<=` | Less than or equal to | `df[df.index <= 1]` |

````python
mask = df.index > 1
````

````python
df
````

````python
df[mask]
````

````python
df.loc[df.index > 1]
````
